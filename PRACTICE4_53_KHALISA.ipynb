**Enkapsulasi**

```python
class User:
    def __init__(self, username, email, gender):
        self._username = username # Enkapsulasi dengan menggunakan underscore (_)
        self._email = email
        self._gender = gender

    def get_username(self):
        return self._username

    def set_username(self, username):
        # Validasi atau logika tambahan dapat ditambahkan di sini
        self._username = username

# Contoh penggunaan
user = User("Khalisa Zalfa Salzabila", "khalisazalfasalsabila@gmail.com", "perempuan")
print(user.get_username()) # Mengakses melalui metode getter
user.set_username("Syaabila.aaa") # Mengubah melalui metode setter
print(user.get_username())
```

**Hands On**

```python
class User:
    def __init__(self, username, email, gender):
        self._username = username
        self._email = email
        self._gender = gender

    def get_gender(self):
        return self._gender

    def set_gender(self, gender):
        # Validasi atau logika tambahan dapat ditambahkan di sini
        self._gender = gender

user = User("Khalisa Zalfa Salzabila", "khalisazalfasalsabila@gmail.com", "perempuan")
print(user.get_gender()) # Mengakses melalui metode getter
user.set_gender("unknown") # Mengubah melalui metode setter
print(user.get_gender())
```

**Inheritance**

```python
class PremiumUser(User):
    def __init__(self, username, email, gender, premium_level):
        super().__init__(username, email, gender)
        self.premium_level = premium_level

    def display_premium_info(self):
        print("Premium Level:", self.premium_level)

premium_user = PremiumUser("Khalisa Zalfa Ssalzabila", "khalisazalfasalsabila@gmail.com", "Perempuan", "Gold")
premium_user.display_premium_info()
```

**Polymorphism**

```python
class User:
    def __init__(self, username, email, gender):
        self.username = username
        self.email = email
        self.gender = gender

    def display_user_info(self):
        print("Username:", self.username)
        print("Email   :", self.email)
        print("Gender  :", self.gender)

class PremiumUser(User):
    def __init__(self, username, email, gender, premium_level):
        # Memanggil init dari class induk (User)
        super().__init__(username, email, gender)
        self.premium_level = premium_level

    def display_user_info(self):
        # Override metode display_user_info
        super().display_user_info()
        print("Premium Level:", self.premium_level)

def display_user_information(user):
    if isinstance(user, User):
        user.display_user_info()
        print("-" * 30) # Garis pembatas agar rapi
    else:
        print("Objek tidak dikenali.")

# --- PERBAIKAN DI SINI ---
# user1 menggunakan class User (3 argumen)
user1 = User("Khalisa Zalfa Salzabila", "khalisazalfasalsabila@gmail.com", "Perempuan")

# user2 menggunakan class PremiumUser (WAJIB 4 argumen: nama, email, gender, level)
user2 = PremiumUser("Khalisa Zalfa Salzabila", "khalisazalfasalsabila@gmail.com", "Perempuan", "Gold")

# Memanggil fungsi untuk menampilkan info
display_user_information(user1)
display_user_information(user2)
```

**Abstraction**

```python
from abc import ABC, abstractmethod

# Kelas abstrak (abstraksi) untuk jenis mobil
class JenisMobil(ABC): # ABC yang diwarisi oleh ABC
    def __init__(self, nama, warna):
        self.nama = nama
        self.warna = warna

    @abstractmethod
    def informasi(self):
        pass

# Kelas turunan yang mengimplementasikan metode abstrak dari JenisMobil
class Sedan(JenisMobil):
    def informasi(self):
        return f"Sedan {self.nama} berwarna {self.warna}"

# Kelas turunan lainnya
class SUV(JenisMobil):
    def informasi(self):
        return f"SUV {self.nama} berwarna {self.warna}"

# Fungsi untuk menampilkan informasi mobil
def tampilkan_informasi(mobil):
    print(mobil.informasi())

# Contoh penggunaan
sedan_mobil = Sedan("Vios", "maroon")
suv_mobil = SUV("CR-V", "putih")

tampilkan_informasi(sedan_mobil)
tampilkan_informasi(suv_mobil)
```
