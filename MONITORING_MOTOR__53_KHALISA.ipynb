```python
import tkinter as tk
from tkinter import ttk

# --- 1. Inisialisasi Utama ---
app = tk.Tk()
app.title("Robot Monitoring System v2.0")
app.geometry("750x600")
app.configure(bg="#1a1a2e") # Background Navy Gelap

# --- 2. Header Style ---
header_bg = tk.Frame(app, bg="#16213e", height=60)
header_bg.pack(fill="x")

title = tk.Label(header_bg, text="DASHBOARD MONITORING ROBOT",
                 font=("Helvetica", 14, "bold"), bg="#3E3089", fg="#6b91c9")
title.pack(pady=15)

# --- Container Tengah ---
container = tk.Frame(app, bg="#1a1a2e")
container.pack(fill="both", expand=True, padx=30, pady=20)

# --- 3. Panel Kiri (Control & Input) ---
panel_kiri = tk.LabelFrame(container, text=" Motor & Servo Control ", font=("Arial", 10, "bold"),
                           bg="#1a1a2e", fg="white", padx=15, pady=15)
panel_kiri.grid(row=0, column=0, sticky="nsew", padx=10)

# Input Rows
def buat_input(parent, teks, baris):
    lbl = tk.Label(parent, text=teks, bg="#1a1a2e", fg="#95a5a6", font=("Arial", 9))
    lbl.grid(row=baris, column=0, sticky="w", pady=5)
    ent = tk.Entry(parent, width=18, bg="#0f3460", fg="white", insertbackground="white", borderwidth=0)
    ent.grid(row=baris, column=1, padx=10, pady=5, ipady=3)
    return ent

e_kanan = buat_input(panel_kiri, "RPM Kanan", 0)
e_kiri  = buat_input(panel_kiri, "RPM Kiri", 1)
e_servo = buat_input(panel_kiri, "Servo Deg", 2)

btn_apply = tk.Button(panel_kiri, text="APPLY SETTINGS", bg="#e94560", fg="white",
                      font=("Arial", 8, "bold"), relief="flat", width=25)
btn_apply.grid(row=3, column=0, columnspan=2, pady=15)

# --- 4. Visualisasi (Canvas Tengah) ---
tk.Label(panel_kiri, text="Posisi Robot (Live)", bg="#1a1a2e", fg="white").grid(row=4, column=0, columnspan=2)
layar_simulasi = tk.Canvas(panel_kiri, width=220, height=180, bg="#0f3460", highlightthickness=1, highlightbackground="#533483")
layar_simulasi.grid(row=5, column=0, columnspan=2, pady=10)
# Aksen robot
layar_simulasi.create_rectangle(100, 80, 120, 100, fill="#e94560", outline="white")

# --- 5. Panel Kanan (Sensors) ---
panel_kanan = tk.LabelFrame(container, text=" Sensor Intelligence ", font=("Arial", 10, "bold"),
                            bg="#1a1a2e", fg="white", padx=15, pady=15)
panel_kanan.grid(row=0, column=1, sticky="nsew", padx=10)

for i in range(1, 6):
    f_sns = tk.Frame(panel_kanan, bg="#1a1a2e")
    f_sns.pack(fill="x", pady=8)

    # Lampu Indikator
    ind = tk.Label(f_sns, text=" ● ", fg="#0fefad", bg="#1a1a2e", font=("Arial", 12))
    ind.pack(side="left")

    tk.Label(f_sns, text=f"Proximity Sensor {i}", bg="#1a1a2e", fg="white", font=("Arial", 9)).pack(side="left")

    # Progress Bar dummy
    pb = ttk.Progressbar(f_sns, length=80, mode='determinate')
    pb['value'] = 40 + (i * 10)
    pb.pack(side="right", padx=5)

btn_refresh = tk.Button(panel_kanan, text="REFRESH SENSOR", bg="#0fefad", fg="#1a1a2e",
                        font=("Arial", 8, "bold"), relief="flat")
btn_refresh.pack(fill="x", pady=20)

# --- 6. Identitas (Footer) ---
footer = tk.Frame(app, bg="#16213e", height=30)
footer.pack(side="bottom", fill="x")
tk.Label(footer, text="Created by: [Khalisa Zalfa Salzabila] | [4222511053] | Robotics PBL 2026",
         font=("Courier", 8), bg="#16213e", fg="#95a5a6").pack(pady=5)

app.mainloop()
```
