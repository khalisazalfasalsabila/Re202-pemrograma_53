```python
import tkinter as tk
from tkinter import ttk
import random # Untuk simulasi deteksi

class SmartTrashFinal:
    def __init__(self, root):
        self.root = root
        self.root.title("Monitoring Smart Trash Bin v2")
        self.root.geometry("750x550")
        self.root.configure(bg="#0f172a")

        # Variabel Data (Untuk menyimpan angka sampah)
        self.count_metal = 0
        self.count_non_metal = 0

        header_frame = tk.Frame(root, bg="#0f172a")
        header_frame.pack(fill="x", padx=30, pady=20)
        tk.Label(header_frame, text="TRACKER PEMILAH SAMPAH", font=("Courier New", 20, "bold"),
                 bg="#0f172a", fg="#38bdf8").pack(side="left")

        content = tk.Frame(root, bg="#0f172a")
        content.pack(fill="both", expand=True, padx=30)

        # --- BAGIAN KIRI: SENSOR ---
        sensor_box = tk.LabelFrame(content, text=" DIAGNOSTIK SENSOR ", font=("Arial", 10, "bold"),
                                  bg="#0f172a", fg="#94a3b8", padx=15, pady=15)
        sensor_box.grid(row=0, column=0, sticky="nsew", padx=(0, 20))

        # Simpan objek canvas sensor agar bisa diubah warnanya
        self.canvas_ir = self.add_sensor(sensor_box, "Infrared Active")
        self.canvas_ind = self.add_sensor(sensor_box, "Induktif (Metal)")
        self.canvas_cap = self.add_sensor(sensor_box, "Kapasitif (Non-Metal)")

        # HUBUNGKAN TOMBOL REFRESH KE FUNGSI SIMULASI
        tk.Button(sensor_box, text="REFRESH SENSOR", bg="#0ea5e9", fg="white",
                  font=("Arial", 9, "bold"), bd=0, height=2,
                  command=self.simulasi_deteksi).pack(pady=20, fill="x")

        # --- BAGIAN KANAN: DATA ---
        data_frame = tk.Frame(content, bg="#0f172a")
        data_frame.grid(row=0, column=1, sticky="nsew")

        # Kita simpan Entry-nya ke dalam variabel agar bisa diupdate isinya
        self.entry_metal = self.add_stat(data_frame, "Total Sampah Logam", "0", 0)
        self.entry_non = self.add_stat(data_frame, "Total Sampah Non-Logam", "0", 1)
        self.entry_servo = self.add_stat(data_frame, "Posisi Servo (Deg)", "90", 2)

        tk.Label(data_frame, text="Simulasi Area Penampungan", bg="#0f172a", fg="#94a3b8", font=("Arial", 9)).grid(row=3, column=0, columnspan=2, pady=(20,5))
        self.canvas = tk.Canvas(data_frame, width=300, height=180, bg="#1e293b", highlightthickness=1, highlightbackground="#334155")
        self.canvas.grid(row=4, column=0, columnspan=3)
        self.rect = self.canvas.create_rectangle(120, 70, 180, 110, outline="#38bdf8", width=2)

    # --- LOGIKA CORONG / DETEKSI (BARU) ---
    def simulasi_deteksi(self):
        """Fungsi ini adalah 'Otak' yang mendeteksi jenis sampah dan menggerakkan corong"""
        jenis = random.choice(["LOGAM", "NON-LOGAM"]) # Simulasi deteksi sensor

        # Reset semua lampu jadi merah dulu
        self.canvas_ir.itemconfig(self.canvas_ir.dot_id, fill="#ef4444")
        self.canvas_ind.itemconfig(self.canvas_ind.dot_id, fill="#ef4444")
        self.canvas_cap.itemconfig(self.canvas_cap.dot_id, fill="#ef4444")

        # Lampu IR selalu nyala saat ada benda
        self.canvas_ir.itemconfig(self.canvas_ir.dot_id, fill="#22c55e")

        if jenis == "LOGAM":
            self.count_metal += 1
            self.canvas_ind.itemconfig(self.canvas_ind.dot_id, fill="#22c55e") # Lampu Induktif Hijau
            self.update_ui(self.entry_metal, self.count_metal)
            self.update_ui(self.entry_servo, "45") # CORONG TERBUKA KE ARAH LOGAM (45 derajat)
        else:
            self.count_non_metal += 1
            self.canvas_cap.itemconfig(self.canvas_cap.dot_id, fill="#22c55e") # Lampu Kapasitif Hijau
            self.update_ui(self.entry_non, self.count_non_metal)
            self.update_ui(self.entry_servo, "135") # CORONG TERBUKA KE ARAH NON-LOGAM (135 derajat)

    def update_ui(self, entry, value):
        entry.delete(0, tk.END)
        entry.insert(0, str(value))

    def add_sensor(self, parent, name):
        f = tk.Frame(parent, bg="#0f172a")
        f.pack(fill="x", pady=8)
        dot_canvas = tk.Canvas(f, width=15, height=15, bg="#0f172a", highlightthickness=0)
        dot_canvas.pack(side="left")
        # Simpan ID lingkaran agar warnanya bisa diganti
        dot_canvas.dot_id = dot_canvas.create_oval(2, 2, 13, 13, fill="#ef4444")
        tk.Label(f, text=name, bg="#0f172a", fg="#f1f5f9", font=("Arial", 10)).pack(side="left", padx=10)
        return dot_canvas

    def add_stat(self, parent, txt, val, r):
        tk.Label(parent, text=txt, bg="#0f172a", fg="#f8fafc", font=("Arial", 11)).grid(row=r, column=0, sticky="w", pady=10)
        e = tk.Entry(parent, width=10, font=("Arial", 11), bg="#1e293b", fg="white", insertbackground="white", bd=1)
        e.insert(0, val)
        e.grid(row=r, column=1, padx=15)
        tk.Button(parent, text="SET", bg="#64748b", fg="white", font=("Arial", 8), width=5).grid(row=r, column=2)
        return e # Kembalikan objek Entry agar bisa diedit datanya

if __name__ == "__main__":
    root = tk.Tk()
    app = SmartTrashFinal(root)
    root.mainloop()
```
