```python
class Karyawan:
    def __init__(self, nama, gaji_pokok):
        self.nama = nama
        # ENCAPSULATION: Menggunakan __ (double underscore) untuk private attribute
        self.__gaji_pokok = gaji_pokok

    # Getter untuk mengakses gaji pokok yang diproteksi
    def get_gaji_pokok(self):
        return self.__gaji_pokok

    def hitung_gaji(self):
        # Metode ini akan di-override oleh class anak
        return self.__gaji_pokok

# INHERITANCE: Manager mewarisi class Karyawan
class Manager(Karyawan):
    def __init__(self, nama, gaji_pokok, bonus):
        super().__init__(nama, gaji_pokok)
        self.bonus = bonus

    # POLYMORPHISM: Mengubah cara hitung_gaji bekerja khusus untuk Manager
    def hitung_gaji(self):
        return self.get_gaji_pokok() + self.bonus

# INHERITANCE: Developer mewarisi class Karyawan
class Developer(Karyawan):
    def __init__(self, nama, gaji_pokok, lembur_per_jam, jam_lembur):
        super().__init__(nama, gaji_pokok)
        self.lembur_per_jam = lembur_per_jam
        self.jam_lembur = jam_lembur

    # POLYMORPHISM: Mengubah cara hitung_gaji bekerja khusus untuk Developer
    def hitung_gaji(self):
        return self.get_gaji_pokok() + (self.lembur_per_jam * self.jam_lembur)

# Fungsi untuk mempresentasikan data (Polymorphism dalam fungsi)
def cetak_slip_gaji(karyawan):
    print(f"Nama Karyawan: {karyawan.nama}")
    print(f"Total Gaji   : Rp{karyawan.hitung_gaji():,}")
    print("-" * 25)

# --- EKSEKUSI ---
karyawan1 = Manager("Khalisa", 8000000, 2000000)
karyawan2 = Developer("Zalfa", 6000000, 50000, 10)

print("=== SLIP GAJI PERUSAHAAN ===\n")
cetak_slip_gaji(karyawan1)
cetak_slip_gaji(karyawan2)
```





**Encapsulation (Enkapsulasi):**
Penjelasan:
 Saya menyembunyikan variabel gaji_pokok menggunakan __ agar tidak bisa diubah sembarangan dari luar class. Data hanya bisa diakses dengan aman melalui fungsi get_gaji_pokok().

**Inheritance (Pewarisan):**

Penjelasan:
 Saya membuat class Manager dan Developer sebagai anak dari class Karyawan. Artinya, mereka otomatis punya atribut nama, tanpa saya harus menulis ulang kodenya. Ini membuat kode lebih efisien.

**Polymorphism (Polimorfisme):**

Penjelasan:
 Meskipun keduanya adalah Karyawan, cara mereka menghitung gaji berbeda. Manager pakai bonus, Developer pakai uang lembur. Saya menggunakan nama fungsi yang sama yaitu hitung_gaji(), tapi isinya menyesuaikan peran masing-masing.
