```python
import customtkinter as ctk
import random
import tkinter as tk
import math
import time


# CONFIGURATION & TEMA VISUAL (MODERN DARK THEME)

ctk.set_appearance_mode("Dark")
ctk.set_default_color_theme("blue")

class SistemPemilahSampahPintar(ctk.CTk):
    def __init__(self):
        super().__init__()

        # Inisialisasi Kode Warna Hexadecimal
        self.WARNA_LATAR_UTAMA    = "#090D16"
        self.WARNA_PANEL_KONTRAST = "#111827"
        self.WARNA_AKSEN_CYAN     = "#00F0FF"
        self.WARNA_STATUS_HIJAU   = "#39FF14"
        self.WARNA_ALARM_MERAH    = "#FF0055"
        self.WARNA_TEKS_UTAMA     = "#E2E8F0"
        self.FONT_MONOSPACE       = "Consolas"

        # Pengaturan Jendela Aplikasi Utama
        self.title("SMART TRASH SORTING BIN - ADVANCED MANUAL MODE")
        self.geometry("1100x650")
        self.resizable(False, False)
        self.configure(fg_color=self.WARNA_LATAR_UTAMA)

        # --- Manajemen Variabel Kontrol & Data State ---
        self.id_animasi_aktif = None
        self.is_mode_otomatis = True
        self.daftar_tombol_manual = []
        self.BATAS_KAPASITAS_MAKSIMAL = 10
        self.is_lampu_kedip = False
        self.waktu_mulai_sistem = time.time()
        self.kategori_sampah_aktif = "KOSONG"


        # TATA LETAK GRID UTAMA

        self.bingkai_utama = ctk.CTkFrame(self, fg_color="transparent")
        self.bingkai_utama.pack(fill="both", expand=True, padx=20, pady=(20, 10))

        self.bingkai_utama.grid_columnconfigure(0, weight=0, minsize=340)
        self.bingkai_utama.grid_columnconfigure(1, weight=1)
        self.bingkai_utama.grid_rowconfigure(0, weight=1)


        # SEKTOR KIRI: PANEL MONITOR STATUS & INPUT DATA PARAMETER

        self.kolom_kiri = ctk.CTkFrame(self.bingkai_utama, fg_color="transparent")
        self.kolom_kiri.grid(row=0, column=0, padx=(0, 15), sticky="nsew")

        # 1. Status Operasional Mesin
        self.panel_status_mesin = ctk.CTkFrame(self.kolom_kiri, fg_color=self.WARNA_PANEL_KONTRAST, corner_radius=4, border_width=1, border_color="#1E293B")
        self.panel_status_mesin.pack(fill="x", pady=(0, 15), ipady=10)

        self.label_notifikasi = ctk.CTkLabel(self.panel_status_mesin, text="KONDISI SISTEM: NORMAL", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11, weight="bold"), text_color=self.WARNA_STATUS_HIJAU)
        self.label_notifikasi.pack(pady=(8, 2), anchor="w", padx=15)

        self.label_deskripsi_status = ctk.CTkLabel(self.panel_status_mesin, text="MESIN SIAP / STANDBY", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=22, weight="bold"), text_color=self.WARNA_AKSEN_CYAN)
        self.label_deskripsi_status.pack(pady=(0, 5), anchor="w", padx=15)

        # 2. Tabel Pengaturan Parameter
        self.panel_parameter = ctk.CTkFrame(self.kolom_kiri, fg_color=self.WARNA_PANEL_KONTRAST, corner_radius=4, border_width=1, border_color="#1E293B")
        self.panel_parameter.pack(fill="both", expand=True)

        ctk.CTkLabel(self.panel_parameter, text="PENGATURAN PARAMETER", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=13, weight="bold"), text_color=self.WARNA_AKSEN_CYAN).pack(pady=15, anchor="w", padx=15)

        self.grid_tabel_data = ctk.CTkFrame(self.panel_parameter, fg_color="transparent")
        self.grid_tabel_data.pack(fill="x", padx=15, pady=5)
        self.grid_tabel_data.grid_columnconfigure(0, weight=0, minsize=150)
        self.grid_tabel_data.grid_columnconfigure(1, weight=0, minsize=75)
        self.grid_tabel_data.grid_columnconfigure(2, weight=0, minsize=85)

        # Header Tabel
        ctk.CTkLabel(self.grid_tabel_data, text="NAMA DATA", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11, weight="bold"), text_color="#566885", anchor="w").grid(row=0, column=0, pady=(0, 10), sticky="w")
        ctk.CTkLabel(self.grid_tabel_data, text="ANGKA", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11, weight="bold"), text_color="#566885").grid(row=0, column=1, pady=(0, 10))
        ctk.CTkLabel(self.grid_tabel_data, text="AKSI", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11, weight="bold"), text_color="#566885").grid(row=0, column=2, pady=(0, 10), padx=(15, 0))

        # Baris Data 1: Kuantitas Sampah Logam
        ctk.CTkLabel(self.grid_tabel_data, text="Sampah Logam", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=12), text_color=self.WARNA_TEKS_UTAMA, anchor="w").grid(row=1, column=0, pady=10, sticky="w")
        self.input_logam = ctk.CTkEntry(self.grid_tabel_data, width=60, height=28, justify="center", fg_color="#090D16", text_color=self.WARNA_STATUS_HIJAU, border_color="#1E293B", font=(self.FONT_MONOSPACE, 12))
        self.input_logam.insert(0, "0")
        self.input_logam.grid(row=1, column=1, pady=10)

        self.btn_ubah_logam = ctk.CTkButton(self.grid_tabel_data, text="SET", width=55, height=26, fg_color="transparent", hover_color="#1E293B", border_width=1, border_color=self.WARNA_AKSEN_CYAN, font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11), corner_radius=2, text_color=self.WARNA_AKSEN_CYAN, command=lambda: self.eksekusi_override_data("Logam", self.input_logam.get()))
        self.btn_ubah_logam.grid(row=1, column=2, pady=10, padx=(15, 0))
        self.daftar_tombol_manual.append(self.btn_ubah_logam)

        # Baris Data 2: Kuantitas Sampah Non-Logam
        ctk.CTkLabel(self.grid_tabel_data, text="Sampah Non-Logam", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=12), text_color=self.WARNA_TEKS_UTAMA, anchor="w").grid(row=2, column=0, pady=10, sticky="w")
        self.input_non_logam = ctk.CTkEntry(self.grid_tabel_data, width=60, height=28, justify="center", fg_color="#090D16", text_color=self.WARNA_STATUS_HIJAU, border_color="#1E293B", font=(self.FONT_MONOSPACE, 12))
        self.input_non_logam.insert(0, "0")
        self.input_non_logam.grid(row=2, column=1, pady=10)

        self.btn_ubah_non_logam = ctk.CTkButton(self.grid_tabel_data, text="SET", width=55, height=26, fg_color="transparent", hover_color="#1E293B", border_width=1, border_color=self.WARNA_AKSEN_CYAN, font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11), corner_radius=2, text_color=self.WARNA_AKSEN_CYAN, command=lambda: self.eksekusi_override_data("Non-Logam", self.input_non_logam.get()))
        self.btn_ubah_non_logam.grid(row=2, column=2, pady=10, padx=(15, 0))
        self.daftar_tombol_manual.append(self.btn_ubah_non_logam)

        # Baris Data 3: Posisi Orientasi Sudut Servo
        ctk.CTkLabel(self.grid_tabel_data, text="Sudut Servo", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=12), text_color=self.WARNA_TEKS_UTAMA, anchor="w").grid(row=3, column=0, pady=10, sticky="w")
        self.input_servo = ctk.CTkEntry(self.grid_tabel_data, width=60, height=28, justify="center", fg_color="#090D16", text_color=self.WARNA_STATUS_HIJAU, border_color="#1E293B", font=(self.FONT_MONOSPACE, 12))
        self.input_servo.insert(0, "90")
        self.input_servo.grid(row=3, column=1, pady=10)

        self.btn_ubah_servo = ctk.CTkButton(self.grid_tabel_data, text="SET", width=55, height=26, fg_color="transparent", hover_color="#1E293B", border_width=1, border_color=self.WARNA_AKSEN_CYAN, font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11), corner_radius=2, text_color=self.WARNA_AKSEN_CYAN, command=lambda: self.eksekusi_override_data("Servo", self.input_servo.get()))
        self.btn_ubah_servo.grid(row=3, column=2, pady=10, padx=(15, 0))
        self.daftar_tombol_manual.append(self.btn_ubah_servo)


        # SEKTOR KANAN: TELEMETRI DIAGNOSTIK & LIVE MONITORING

        self.kolom_kanan = ctk.CTkFrame(self.bingkai_utama, fg_color="transparent")
        self.kolom_kanan.grid(row=0, column=1, sticky="nsew")

        # 1. Subsistem Lampu Indikator Sensor
        self.panel_sensor_bus = ctk.CTkFrame(self.kolom_kanan, fg_color=self.WARNA_PANEL_KONTRAST, corner_radius=4, border_width=1, border_color="#1E293B")
        self.panel_sensor_bus.pack(fill="x", pady=(0, 15))

        ctk.CTkLabel(self.panel_sensor_bus, text="PANEL INDIKATOR SENSOR ", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11, weight="bold"), text_color="#566885").pack(pady=5)

        self.kontainer_sensor = ctk.CTkFrame(self.panel_sensor_bus, fg_color="transparent")
        self.kontainer_sensor.pack(fill="x", padx=10, pady=(0, 10))
        self.kontainer_sensor.grid_columnconfigure((0, 1, 2), weight=1, uniform="equal")

        self.lampu_ir = self.membangun_widget_indikator_sensor(self.kontainer_sensor, "Sensor IR (Objek)", 0)
        self.lampu_induktif = self.membangun_widget_indikator_sensor(self.kontainer_sensor, "Sensor Induktif (Logam)", 1)
        self.lampu_kapasitif = self.membangun_widget_indikator_sensor(self.kontainer_sensor, "Sensor Kapasitif (Non-Logam)", 2)

        # 2. Mesin Visualisasi Animasi Mekanik Canvas
        self.panel_visualisasi = ctk.CTkFrame(self.kolom_kanan, fg_color=self.WARNA_PANEL_KONTRAST, corner_radius=4, border_width=1, border_color="#1E293B")
        self.panel_visualisasi.pack(fill="both", expand=True, pady=(0, 15))

        ctk.CTkLabel(self.panel_visualisasi, text="VISUALISASI GERAKAN MEKANIK & SERVO", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=12, weight="bold"), text_color=self.WARNA_TEKS_UTAMA).pack(pady=10)

        self.monitor_canvas = tk.Canvas(self.panel_visualisasi, width=700, height=210, bg="#060911", highlightthickness=0)
        self.monitor_canvas.pack(pady=5, padx=15,expand=True)

        self.titik_tengah_x = 350

        # Grid Latar Belakang (Desain Grid Cyberpunk)
        for horizontal_x in range(0, 1100, 20): self.monitor_canvas.create_line(horizontal_x, 0, horizontal_x, 210, fill="#0F1626", width=1)
        for vertikal_y in range(0, 410, 20): self.monitor_canvas.create_line(0, vertikal_y, 700, vertikal_y, fill="#0F1626", width=1)

        # Struktur Katup Gerbang Atas
        self.monitor_canvas.create_rectangle(self.titik_tengah_x - 100, 15, self.titik_tengah_x + 100, 25, outline="#1E293B", fill="#0D1527")
        self.pintu_kiri = self.monitor_canvas.create_rectangle(self.titik_tengah_x - 100, 16, self.titik_tengah_x, 24, outline="", fill=self.WARNA_AKSEN_CYAN)
        self.pintu_kanan = self.monitor_canvas.create_rectangle(self.titik_tengah_x, 16, self.titik_tengah_x + 100, 24, outline="", fill=self.WARNA_AKSEN_CYAN)

        # Batas Geometris Deteksi Sensor
        self.monitor_canvas.create_line(40, 75, 660, 75, fill=self.WARNA_ALARM_MERAH, dash=(6, 4), width=1.5)
        self.monitor_canvas.create_text(self.titik_tengah_x, 63, text="GARIS BATAS DETEKSI SENSOR", fill=self.WARNA_ALARM_MERAH, font=(self.FONT_MONOSPACE, 8, "bold"))

        # Geometri Lengan Pengarah Servo Motor
        self.servo_poros_x, self.servo_poros_y = self.titik_tengah_x, 115
        self.panjang_lengan_servo = 55
        self.monitor_canvas.create_rectangle(self.servo_poros_x - 15, 100, self.servo_poros_x + 15, 115, fill="#1E293B", outline=self.WARNA_AKSEN_CYAN)
        self.garis_vektor_servo = self.monitor_canvas.create_line(self.servo_poros_x, self.servo_poros_y, self.servo_poros_x, self.servo_poros_y + self.panjang_lengan_servo, fill="#FFCC00", width=5, capstyle="round")

        # Bak Penampung Hasil Pemilahan
        self.bak_logam = self.monitor_canvas.create_rectangle(self.titik_tengah_x - 160, 150, self.titik_tengah_x - 80, 200, outline=self.WARNA_STATUS_HIJAU, width=1.5, fill="#0D1F1D")
        self.monitor_canvas.create_text(self.titik_tengah_x - 120, 175, text="BIN\nLOGAM", fill=self.WARNA_STATUS_HIJAU, font=(self.FONT_MONOSPACE, 9, "bold"))

        self.bak_non_logam = self.monitor_canvas.create_rectangle(self.titik_tengah_x + 80, 150, self.titik_tengah_x + 160, 200, outline=self.WARNA_AKSEN_CYAN, width=1.5, fill="#0A212E")
        self.monitor_canvas.create_text(self.titik_tengah_x + 120, 175, text="BIN\nNON-LOGAM", fill=self.WARNA_AKSEN_CYAN, font=(self.FONT_MONOSPACE, 8, "bold"))

        # Payload Object (Bola Sampah)
        self.objek_sampah_bola = self.monitor_canvas.create_oval(self.titik_tengah_x - 6, -20, self.titik_tengah_x + 6, -10, fill="#FFFFFF", outline="")

        # 3. Log System Monitor & Pengisian Panel Samping Kosong (DETEKSI)
        self.sub_bingkai_log = ctk.CTkFrame(self.kolom_kanan, fg_color="transparent")
        self.sub_bingkai_log.pack(fill="x", pady=(0, 15))

        # --- DISINI PERBAIKAN PANEL SAMPING KOSONG ---
        self.panel_kategori = ctk.CTkFrame(self.sub_bingkai_log, fg_color=self.WARNA_PANEL_KONTRAST, corner_radius=4, border_width=1, border_color="#1E293B", width=180, height=110)
        self.panel_kategori.pack(side="left", fill="y", padx=(0, 10))
        self.panel_kategori.pack_propagate(False)
        ctk.CTkLabel(self.panel_kategori, text="MATRIKS DETEKSI:", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=9, weight="bold"), text_color="#566885").pack(pady=(15, 2))

        # Menampilkan teks besar isi deteksi jenis material terakhir
        self.label_hasil_klasifikasi = ctk.CTkLabel(self.panel_kategori, text="READY", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=15, weight="bold"), text_color="#4B5563")
        self.label_hasil_klasifikasi.pack()

        # Menampilkan status info sudut servo saat ini di bawah tulisan deteksi
        self.label_info_servo = ctk.CTkLabel(self.panel_kategori, text="SERVO: 90°", font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=11), text_color="#566885")
        self.label_info_servo.pack(pady=(5, 0))

        self.kotak_log_aktivitas = ctk.CTkTextbox(self.sub_bingkai_log, font=ctk.CTkFont(family="Consolas", size=10), height=110, fg_color="#060911", border_width=1, border_color="#1E293B", text_color=self.WARNA_AKSEN_CYAN)
        self.kotak_log_aktivitas.pack(side="left", fill="both", expand=True)
        self.kotak_log_aktivitas.insert("0.0", "[SISTEM] Core TIM_3Berhasil Diinisialisasi...\n")
        self.kotak_log_aktivitas.configure(state="disabled")

        # 4. Bus Tombol Navigasi Operasional
        self.baris_tombol_aksi = ctk.CTkFrame(self.kolom_kanan, fg_color="transparent")
        self.baris_tombol_aksi.pack(fill="x")

        self.btn_trigger_simulasi = ctk.CTkButton(
            self.baris_tombol_aksi, text="Mulai Simulasi Pemrosesan Sampah", height=42, fg_color="#1E3A8A", hover_color="#2563EB",
            font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=12, weight="bold"), text_color=self.WARNA_AKSEN_CYAN, corner_radius=2, command=self.proses_automasi_pemilahan
        )
        self.btn_trigger_simulasi.pack(side="left", fill="x", expand=True, padx=(0, 5))

        self.btn_pengubah_mode = ctk.CTkButton(
            self.baris_tombol_aksi, text="Mode: OTOMATIS", height=42, width=160, fg_color="transparent",
            text_color=self.WARNA_STATUS_HIJAU, border_width=1, border_color=self.WARNA_STATUS_HIJAU, font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=12, weight="bold"), corner_radius=2, command=self.beralih_mode_kendali
        )
        self.btn_pengubah_mode.pack(side="left", padx=(0, 5))

        self.btn_reset_storage = ctk.CTkButton(
            self.baris_tombol_aksi, text="RESET", fg_color="transparent", hover_color="#31101D",
            text_color=self.WARNA_ALARM_MERAH, border_width=1, border_color=self.WARNA_ALARM_MERAH, font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=12, weight="bold"), height=42, width=140, corner_radius=2, command=self.reset_penyimpanan_sampah
        )
        self.btn_reset_storage.pack(side="left")

        # --- PANEL FOOTER ---
        self.status_bar_footer = ctk.CTkFrame(self, fg_color="transparent")
        self.status_bar_footer.pack(fill="x", side="bottom", padx=20, pady=(5, 10))

        self.label_spesifikasi_cpu = ctk.CTkLabel(self.status_bar_footer, text="KONEKSI DATA BUS: ELEKTRONIK AKTIF  //  BEBAN PROSESOR: 14.2%  //  FRAME RATE: 60 FPS", font=(self.FONT_MONOSPACE, 10), text_color="#475569")
        self.label_spesifikasi_cpu.pack(side="left")

        self.label_durasi_aktif = ctk.CTkLabel(self.status_bar_footer, text="DURASI OPERASIONAL: 00:00:00", font=(self.FONT_MONOSPACE, 10), text_color="#475569")
        self.label_durasi_aktif.pack(side="right")

        self.sinkronisasi_ui_mode_operasi()
        self.jalankan_loop_proteksi_kapasitas()
        self.jalankan_hitungan_waktu_nyala()

    # =================================================================
    # SUBSISTEM LOGIKA KINEMATIK SERVO & TELEMETRI
    # =================================================================
    def jalankan_hitungan_waktu_nyala(self):
        waktu_berjalan = int(time.time() - self.waktu_mulai_sistem)
        jam = waktu_berjalan // 3600
        menit = (waktu_berjalan % 3600) // 60
        detik = waktu_berjalan % 60
        self.label_durasi_aktif.configure(text=f"DURASI OPERASIONAL: {jam:02d}:{menit:02d}:{detik:02d}")
        self.after(1000, self.jalankan_hitungan_waktu_nyala)

    def kalkulasi_vektor_sudut_servo(self, target_derajat):
        sudut_radian = math.radians(target_derajat)
        ujung_vektor_x = self.servo_poros_x + self.panjang_lengan_servo * math.cos(sudut_radian)
        ujung_vektor_y = self.servo_poros_y + self.panjang_lengan_servo * math.sin(sudut_radian)
        self.monitor_canvas.coords(self.garis_vektor_servo, self.servo_poros_x, self.servo_poros_y, ujung_vektor_x, ujung_vektor_y)
        self.label_info_servo.configure(text=f"SERVO: {target_derajat}°")

    def membangun_widget_indikator_sensor(self, parent, nama_sensor, indeks_kolom):
        frame_independen = ctk.CTkFrame(parent, fg_color="transparent")
        frame_independen.grid(row=0, column=indeks_kolom, padx=5, pady=2, sticky="nsew")

        lbl = ctk.CTkLabel(frame_independen, text=nama_sensor, font=ctk.CTkFont(family=self.FONT_MONOSPACE, size=10), text_color="#566885")
        lbl.pack(pady=(0, 4))

        kanvas_lampu = tk.Canvas(frame_independen, width=40, height=8, bg=self.WARNA_PANEL_KONTRAST, highlightthickness=0)
        kanvas_lampu.pack()

        bentuk_oval = kanvas_lampu.create_rectangle(0, 0, 40, 8, fill="#1E293B", outline="#2D3748")
        frame_independen.kanvas_lampu = kanvas_lampu
        frame_independen.bentuk_oval = bentuk_oval
        return frame_independen

    def injeksi_warna_sensor(self, komponen_sensor, warna_fill, warna_grid):
        komponen_sensor.kanvas_lampu.itemconfig(komponen_sensor.bentuk_oval, fill=warna_fill, outline=warna_grid)

    def cetak_catatan_sistem(self, payload_pesan):
        self.kotak_log_aktivitas.configure(state="normal")
        self.kotak_log_aktivitas.insert("end", f">> {payload_pesan}\n")
        self.kotak_log_aktivitas.see("end")
        self.kotak_log_aktivitas.configure(state="disabled")

    def perbarui_judul_status_mesin(self, status_kode):
        if status_kode == "PROSES":
            self.label_deskripsi_status.configure(text="PROSES PEMILAHAN...", text_color="#FF9900")
        else:
            self.label_deskripsi_status.configure(text="MESIN SIAP / STANDBY", text_color=self.WARNA_AKSEN_CYAN)

    def beralih_mode_kendali(self):
        self.is_mode_otomatis = not self.is_mode_otomatis
        self.sinkronisasi_ui_mode_operasi()
        string_mode = "OTOMATIS" if self.is_mode_otomatis else "MANUAL USER OVERRIDE"
        self.cetak_catatan_sistem(f"Mode Kendali Sistem Dialihkan Ke: {string_mode}")

    def sinkronisasi_ui_mode_operasi(self):
        if self.is_mode_otomatis:
            self.btn_pengubah_mode.configure(text="Mode: OTOMATIS", text_color=self.WARNA_STATUS_HIJAU, border_color=self.WARNA_STATUS_HIJAU)
            self.btn_trigger_simulasi.configure(state="normal")
            for tombol in self.daftar_tombol_manual: tombol.configure(state="disabled")
        else:
            self.btn_pengubah_mode.configure(text="Mode: MANUAL", text_color="#FF9900", border_color="#FF9900")
            self.btn_trigger_simulasi.configure(state="disabled")
            for tombol in self.daftar_tombol_manual: tombol.configure(state="normal")

    # =================================================================
    # INTERVENSI MODE MANUAL (USER OVERRIDE ACTION)
    # =================================================================
    def eksekusi_override_data(self, target_komponen, data_inputan):
        if self.is_mode_otomatis or data_inputan.strip() == "": return

        if target_komponen == "Servo":
            try:
                self.kalkulasi_vektor_sudut_servo(int(data_inputan))
                if sudut not in [45, 135, 90]:
                    self.cetak_catatan_sistem("[PANDUAN] Gunakan sudut 45 (Non-Logam) atau 135 (Logam) untuk simulasi.")
                    return

                self.cetak_catatan_sistem(f"Manual Override -> Mengatur Sudut Servo Ke: {sudut}°")
                self.kalkulasi_vektor_sudut_servo(sudut)

                # JIKA MANUAL DIUBAH 45 / 135, JALANKAN ANIMASI JATUH SECARA MANUAL!
                if sudut == 135:
                    self.kategori_sampah_aktif = "LOGAM"
                    self.proses_mekanik_manual_drop()
                elif sudut == 45:
                    self.kategori_sampah_aktif = "NON-LOGAM"
                    self.proses_mekanik_manual_drop()
            except ValueError: pass

        elif target_komponen == "Logam":
            self.cetak_catatan_sistem(f"Manual Override -> Register Logam Diubah Menjadi: {data_inputan}")
            self.label_hasil_klasifikasi.configure(text="MAN_LOGAM", text_color=self.WARNA_STATUS_HIJAU)
        elif target_komponen == "Non-Logam":
            self.cetak_catatan_sistem(f"Manual Override -> Register Non-Logam Diubah Menjadi: {data_inputan}")
            self.label_hasil_klasifikasi.configure(text="MAN_N-LOGAM", text_color=self.WARNA_AKSEN_CYAN)

    def proses_mekanik_manual_drop(self):
        """ Menembakkan animasi bola jatuh berdasarkan input manual user """
        if self.id_animasi_aktif: return
        self.perbarui_judul_status_mesin("PROSES")

        # Buka Katup Atas
        self.monitor_canvas.coords(self.pintu_kiri, self.titik_tengah_x - 100, 16, self.titik_tengah_x - 40, 24)
        self.monitor_canvas.coords(self.pintu_kanan, self.titik_tengah_x + 40, 16, self.titik_tengah_x + 100, 24)
        self.monitor_canvas.coords(self.objek_sampah_bola, self.titik_tengah_x - 6, 25, self.titik_tengah_x + 6, 35)

        self.loop_pembaharuan_kinematik_bola_manual(25, self.titik_tengah_x)

    def loop_pembaharuan_kinematik_bola_manual(self, koordinat_y, koordinat_x):
        koordinat_y += 4

        # Nyalakan sensor saat menyentuh sensor line secara manual
        if 65 < koordinat_y < 85:
            self.injeksi_warna_sensor(self.lampu_ir, self.WARNA_STATUS_HIJAU, "#1E3A8A")
            if self.kategori_sampah_aktif == "LOGAM":
                self.injeksi_warna_sensor(self.lampu_induktif, self.WARNA_STATUS_HIJAU, "#1E3A8A")
            else:
                self.injeksi_warna_sensor(self.lampu_kapasitif, self.WARNA_STATUS_HIJAU, "#1E3A8A")
        elif koordinat_y >= 115:
            self.injeksi_warna_sensor(self.lampu_ir, "#2D3748", "#1A202C")
            self.injeksi_warna_sensor(self.lampu_induktif, "#2D3748", "#1A202C")
            self.injeksi_warna_sensor(self.lampu_kapasitif, "#2D3748", "#1A202C")

            # Deviasi trayektori jatuh berdasarkan status register
            if self.kategori_sampah_aktif == "LOGAM":
                koordinat_x -= 5.5
            else:
                koordinat_x += 5.5

        self.monitor_canvas.coords(self.objek_sampah_bola, koordinat_x - 6, koordinat_y - 6, koordinat_x + 6, koordinat_y + 6)

        if koordinat_y < 165:
            self.id_animasi_aktif = self.after(15, lambda: self.loop_pembaharuan_kinematik_bola_manual(koordinat_y, koordinat_x))
        else:
            self.konklusi_akhir_sorting_manual()

    def konklusi_akhir_sorting_manual(self):
        # Tutup Katup Kembali
        self.monitor_canvas.coords(self.pintu_kiri, self.titik_tengah_x - 100, 16, self.titik_tengah_x, 24)
        self.monitor_canvas.coords(self.pintu_kanan, self.titik_tengah_x, 16, self.titik_tengah_x + 100, 24)

        if self.kategori_sampah_aktif == "LOGAM":
            angka_skrg = int(self.input_logam.get())
            self.input_logam.delete(0, "end")
            self.input_logam.insert(0, str(angka_skrg + 1))
            self.label_hasil_klasifikasi.configure(text="MAN_LOGAM", text_color=self.WARNA_STATUS_HIJAU)
            self.cetak_catatan_sistem("Manual Execution Berhasil: Objek diarahkan ke BIN LOGAM.")
        else:
            angka_skrg = int(self.input_non_logam.get())
            self.input_non_logam.delete(0, "end")
            self.input_non_logam.insert(0, str(angka_skrg + 1))
            self.label_hasil_klasifikasi.configure(text="MAN_N-LOGAM", text_color=self.WARNA_AKSEN_CYAN)
            self.cetak_catatan_sistem("Manual Execution Berhasil: Objek diarahkan ke BIN NON-LOGAM.")

        self.after(200, lambda: self.monitor_canvas.coords(self.objek_sampah_bola, self.titik_tengah_x - 6, -20, self.titik_tengah_x + 6, -10))
        self.after(600, self.kembalikan_poros_servo_standby)

    # =================================================================
    # LOOP AUTOMATION ENGINE (MODE OTOMATIS)
    # =================================================================
    def jalankan_loop_proteksi_kapasitas(self):
        try:
            total_logam = int(self.input_logam.get())
            total_non_logam = int(self.input_non_logam.get())
        except ValueError:
            total_logam, total_non_logam = 0, 0

        if total_logam >= self.BATAS_KAPASITAS_MAKSIMAL or total_non_logam >= self.BATAS_KAPASITAS_MAKSIMAL:
            self.is_lampu_kedip = not self.is_lampu_kedip
            warna_alert = self.WARNA_ALARM_MERAH if self.is_lampu_kedip else self.WARNA_PANEL_KONTRAST
            self.label_notifikasi.configure(text="⚠ PERINGATAN CRITICAL: STORAGE CAPACITY MAX LIMIT! ⚠", text_color=warna_alert)
            self.btn_trigger_simulasi.configure(state="disabled")
        else:
            self.label_notifikasi.configure(text="KONDISI SISTEM: NORMAL", text_color=self.WARNA_STATUS_HIJAU)
            if self.is_mode_otomatis and not self.id_animasi_aktif:
                self.btn_trigger_simulasi.configure(state="normal")

        self.after(400, self.jalankan_loop_proteksi_kapasitas)

    def reset_penyimpanan_sampah(self):
        self.input_logam.delete(0, "end")
        self.input_logam.insert(0, "0")
        self.input_non_logam.delete(0, "end")
        self.input_non_logam.insert(0, "0")
        self.label_hasil_klasifikasi.configure(text="READY", text_color="#4B5563")
        self.cetak_catatan_sistem("Tindakan: Membersihkan Register & Mengosongkan Seluruh Bak Sampah.")

    def proses_automasi_pemilahan(self):
        if self.id_animasi_aktif or not self.is_mode_otomatis: return
        self.perbarui_judul_status_mesin("PROSES")

        self.kategori_sampah_aktif = random.choice(["LOGAM", "NON-LOGAM"])
        self.cetak_catatan_sistem("Mode Otomatis: Objek terdeteksi. Memulai scanning otomatis...")

        self.monitor_canvas.coords(self.pintu_kiri, self.titik_tengah_x - 100, 16, self.titik_tengah_x - 40, 24)
        self.monitor_canvas.coords(self.pintu_kanan, self.titik_tengah_x + 40, 16, self.titik_tengah_x + 100, 24)
        self.monitor_canvas.coords(self.objek_sampah_bola, self.titik_tengah_x - 6, 25, self.titik_tengah_x + 6, 35)

        self.loop_pembaharuan_kinematik_bola(25, self.titik_tengah_x)

    def loop_pembaharuan_kinematik_bola(self, koordinat_y, koordinat_x):
        koordinat_y += 4

        if koordinat_y < 115:
            if 65 < koordinat_y < 85:
                self.injeksi_warna_sensor(self.lampu_ir, self.WARNA_STATUS_HIJAU, "#1E3A8A")
                if self.kategori_sampah_aktif == "LOGAM":
                    self.injeksi_warna_sensor(self.lampu_induktif, self.WARNA_STATUS_HIJAU, "#1E3A8A")
                    self.input_servo.delete(0, "end")
                    self.input_servo.insert(0, "135")
                    self.kalkulasi_vektor_sudut_servo(135)
                else:
                    self.injeksi_warna_sensor(self.lampu_kapasitif, self.WARNA_STATUS_HIJAU, "#1E3A8A")
                    self.input_servo.delete(0, "end")
                    self.input_servo.insert(0, "45")
                    self.kalkulasi_vektor_sudut_servo(45)
        else:
            self.injeksi_warna_sensor(self.lampu_ir, "#2D3748", "#1A202C")
            self.injeksi_warna_sensor(self.lampu_induktif, "#2D3748", "#1A202C")
            self.injeksi_warna_sensor(self.lampu_kapasitif, "#2D3748", "#1A202C")

            if self.kategori_sampah_aktif == "LOGAM":
                koordinat_x -= 5.5
            else:
                koordinat_x += 5.5

        self.monitor_canvas.coords(self.objek_sampah_bola, koordinat_x - 6, koordinat_y - 6, koordinat_x + 6, koordinat_y + 6)

        if koordinat_y < 165:
            self.id_animasi_aktif = self.after(15, lambda: self.loop_pembaharuan_kinematik_bola(koordinat_y, koordinat_x))
        else:
            self.konklusi_akhir_sorting()

    def konklusi_akhir_sorting(self):
        self.monitor_canvas.coords(self.pintu_kiri, self.titik_tengah_x - 100, 16, self.titik_tengah_x, 24)
        self.monitor_canvas.coords(self.pintu_kanan, self.titik_tengah_x, 16, self.titik_tengah_x + 100, 24)

        if self.kategori_sampah_aktif == "LOGAM":
            angka_skrg = int(self.input_logam.get())
            self.input_logam.delete(0, "end")
            self.input_logam.insert(0, str(angka_skrg + 1))
            self.label_hasil_klasifikasi.configure(text="AUTO_LOGAM", text_color=self.WARNA_STATUS_HIJAU)
            self.cetak_catatan_sistem("Sort Otomatis Berhasil: Klasifikasi LOGAM (Servo 135°)")
        else:
            angka_skrg = int(self.input_non_logam.get())
            self.input_non_logam.delete(0, "end")
            self.input_non_logam.insert(0, str(angka_skrg + 1))
            self.label_hasil_klasifikasi.configure(text="AUTO_N-LOGAM", text_color=self.WARNA_AKSEN_CYAN)
            self.cetak_catatan_sistem("Sort Otomatis Berhasil: Klasifikasi NON-LOGAM (Servo 45°)")

        self.after(200, lambda: self.monitor_canvas.coords(self.objek_sampah_bola, self.titik_tengah_x - 6, -20, self.titik_tengah_x + 6, -10))
        self.after(600, self.kembalikan_poros_servo_standby)

    def kembalikan_poros_servo_standby(self):
        self.input_servo.delete(0, "end")
        self.input_servo.insert(0, "90")
        self.kalkulasi_vektor_sudut_servo(90)
        self.perbarui_judul_status_mesin("STOP")
        self.id_animasi_aktif = None

if __name__ == "__main__":
    aplikasi_pemilah = SistemPemilahSampahPintar()
    aplikasi_pemilah.mainloop()

```
